# 尝试从binance公开数据库下载数据并且转换为qlib格式

+ 以 1min 数据为例进行尝试
+ 获取binance BTC-USDT
+ 数据分为daily更新和monthly更新，从daily下载

```python
from binance_batch_downloader import BinanceBatchDownloader

batch_downloader = BinanceBatchDownloader(
    pairs=['BTCUSDT', 'ETHUSDT'],
    download_path="binance_data_1min",
    start_date='2025-06-01',
    overwrite_mode=False,
    enable_logging=True 
)

# 执行下载
batch_downloader.download_all()

# 获取失败记录
failed_records = batch_downloader.get_failed_downloads()
print("详细失败记录:", failed_records)
```
